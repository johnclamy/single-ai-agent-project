# Single-Agent Experimentation

Prototype a LangChain agent using Gemini and Tavily Search. The cells below configure the agent and run a current-events query.

```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_tavily import TavilySearch

load_dotenv('../.env')
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0)
tools = [TavilySearch(max_results=2)]
agent_executor = create_agent(
    model=llm,
    tools=tools,
    system_prompt=(
        "You are a helpful assistant. Use Tavily Search when current or "
        "up-to-date information is needed."
    ),
)
```

```python
query = "What are the latest advancements in quantum computing announced this week?"
result = agent_executor.invoke({"messages": [{"role": "user", "content": query}]})

print("\n--- FINAL ANSWER ---")
print(result["messages"][-1].content)
```
